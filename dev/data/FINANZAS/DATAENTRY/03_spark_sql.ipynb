```python
data = [
    (1, "Juan", "Lima", 2500),
    (2, "Pedro", "Arequipa", 3200),
    (3, "Maria", "Lima", 2800),
    (4, "Ana", "Cusco", 3500),
    (5, "Luis", "Lima", 1800)
]

columns = ["id", "nombre", "ciudad", "salario"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
df.createOrReplaceTempView("empleados")
```

```python
resultado = spark.sql("""
    SELECT
        ciudad,
        COUNT(*) AS cantidad_empleados,
        ROUND(AVG(salario), 2) AS salario_promedio
    FROM empleados
    GROUP BY ciudad
    ORDER BY salario_promedio DESC
""")

display(resultado)
```

```python
resultado = spark.sql("""
    SELECT *
    FROM empleados
    WHERE salario >= 2500
    ORDER BY salario DESC
""")

display(resultado)
```
