```python
import pandas as pd

data = {
    "cliente_id": [1, 2, 3, 4, 5],
    "segmento": ["A", "B", "A", "C", "B"],
    "saldo": [1200.50, 340.00, 980.75, 15.20, 2500.00],
    "dias_mora": [0, 15, 0, 60, 5],
}
df = pd.DataFrame(data)
df
```

```python
df["en_mora"] = df["dias_mora"] > 30

resumen = (
    df.groupby("segmento")
      .agg(clientes=("cliente_id", "count"),
           saldo_total=("saldo", "sum"),
           mora_max=("dias_mora", "max"))
      .reset_index()
)
resumen
```

```python
assert df["cliente_id"].is_unique, "cliente_id duplicado"
assert df["saldo"].ge(0).all(), "saldo negativo"
print(f"Filas: {len(df)} | En mora: {df['en_mora'].sum()}")
```
