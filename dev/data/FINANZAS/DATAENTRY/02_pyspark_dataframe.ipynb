```python
data = [
    (1, "Juan", "Lima", 2500),
    (2, "Pedro", "Arequipa", 3200),
    (3, "Maria", "Lima", 2800),
    (4, "Ana", "Cusco", 3500),
    (5, "Luis", "Lima", 1800)
]

columns = ["id", "nombre", "ciudad", "salario"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
df_lima = df.filter(df.ciudad == "Lima")

display(df_lima)
```

```python
from pyspark.sql.functions import avg, count

df_resumen = df.groupBy("ciudad").agg(
    count("*").alias("cantidad_empleados"),
    avg("salario").alias("salario_promedio")
)

display(df_resumen)
```
